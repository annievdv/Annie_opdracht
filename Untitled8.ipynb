```python
import numpy as np
import pandas as pd
from google.colab import files

#print("Please upload the file now. A file browser window will appear.")
#uploaded = files.upload()

# Assuming the correct file was uploaded, it will now be in the /content/ directory.
# The existing path for pd.read_csv should work after the upload.
df=pd.read_csv('https://raw.githubusercontent.com/annievdv/Annie_opdracht/refs/heads/main/per-capita-energy-use.csv')

display(df)
```

```python
import matplotlib.pyplot as plt

# We want to see the entire rows with the largest values in the '2021' column
# First, identify the top 5 entities based on 'Fossil fuels per capita' in 2021
df_filtered = df[df['Entity'].str.contains("World|European Union") == False].reset_index(drop=True)
top_10_entities_2024 = df_filtered[df_filtered['Year'] == 2024].sort_values(by='Per capita energy consumption', ascending=False).head(10)['Entity'].tolist()

# Then, filter the DataFrame for these top 5 entities across all years
df_top_10 = df_filtered[df_filtered['Entity'].isin(top_10_entities_2024)]

# Pivot the DataFrame to have 'Entity' as index, 'Year' as columns, and 'per-capita-energy-use' as values
most_energy = df_top_10.pivot(index='Entity', columns='Year', values='Per capita energy consumption')

# Plot
plt.figure(figsize=[10,8])

for index, row in most_energy.iterrows():
    plt.plot(row.index, row.values, marker='o', label=index)


plt.title('Energy')
plt.xlabel('Year')
plt.ylabel('Energy')
plt.legend()
plt.grid()
plt.show()
```
