```python
import numpy as np
import pandas as pd
from google.colab import files

df_energy=pd.read_csv('https://raw.githubusercontent.com/annievdv/Annie_opdracht/refs/heads/main/per-capita-energy-use.csv')
df_gdp=pd.read_csv('https://raw.githubusercontent.com/annievdv/Annie_opdracht/refs/heads/main/gdp-per-capita-maddison-project-database.csv')
df_co2=pd.read_csv('https://raw.githubusercontent.com/annievdv/Annie_opdracht/refs/heads/main/owid-co2-data.csv')

display(df_energy)
display(df_gdp)
display(df_co2)
```

```python
import matplotlib.pyplot as plt

df_filtered_energy = df_energy[df_energy['Entity'].str.contains("World|European Union") == False].reset_index(drop=True)
top_10_entities_2024_energy = df_filtered_energy[df_filtered_energy['Year'] == 2024].sort_values(by='Per capita energy consumption', ascending=False).head(10)['Entity'].tolist()

df_filtered_co2 = df_co2[df_co2['country'].str.contains("World|European Union") == False].reset_index(drop=True)
top_10_entities_2024_co2 = df_filtered_co2[df_filtered_co2['year'] == 2024].sort_values(by='co2_per_capita', ascending=False).head(10)['country'].tolist()

df_filtered_gdp = df_gdp[df_gdp['Entity'].str.contains("World|European Union") == False].reset_index(drop=True)
top_10_entities_2024_gdp = df_filtered_gdp[df_filtered_gdp['Year'] == 2024].sort_values(by='GDP per capita', ascending=False).head(10)['Entity'].tolist()

# Then, filter the DataFrame for these top 10 entities across all years
df_top_10_energy = df_filtered_energy[df_filtered_energy['Entity'].isin(top_10_entities_2024_energy)]
df_top_10_co2 = df_filtered_co2[df_filtered_co2['country'].isin(top_10_entities_2024_co2)]
df_top_10_gdp = df_filtered_gdp[df_filtered_gdp['Entity'].isin(top_10_entities_2024_gdp)]


# Pivot the DataFrame to have 'Entity' as index, 'Year' as columns, and 'per-capita-energy-use' as values
most_energy = df_top_10_energy.pivot(index='Entity', columns='Year', values='Per capita energy consumption')
most_gdp = df_top_10_energy.pivot(index='Entity', columns='Year', values='Per capita energy consumption')

# Plot
plt.figure(figsize=[10,8])

for index, row in most_energy.iterrows():
    plt.plot(row.index, row.values, marker='o', label=index)
for index, row in most_gdp.iterrows():
    plt.plot(row.index, row.values, marker='o', label=index)


# Create scatter plot
plt.scatter(most_energy, most_gdp, color='blue', marker='o')

# Add labels and title
plt.xlabel('Energy')
plt.ylabel('Gdp')
plt.title('Energy versus Gdp')

# Show plot
plt.show()
```
