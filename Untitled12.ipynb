```python
import numpy as np
import pandas as pd
from google.colab import files

#print("Please upload the file now. A file browser window will appear.")
uploaded = files.upload()

# Lees de fossiele brandstoffen data
fossil_fuels_df = pd.read_csv("/content/fossil-fuels-per-capita.csv")
print("Fossiele brandstoffen data (eerste 5 rijen):")
display(fossil_fuels_df.head())

#print("Please upload the file now. A file browser window will appear.")
uploaded = files.upload()

# Lees de bevolkingsdata
population_df = pd.read_csv("/content/population-with-un-projections.csv")
print("\nBevolkingsdata (eerste 5 rijen):")
display(population_df.head())
```

```python
# Aangenomen dat fossil_fuels_df en population_df nu correct zijn gedefinieerd in de vorige cel.
# Samenvoegen op gemeenschappelijke kolommen 'Entity' en 'Year'.
merged_df = pd.merge(fossil_fuels_df, population_df, on=['Entity', 'Year'], how='outer')
print("Samengevoegde dataframe (eerste 5 rijen):")
display(merged_df.head())

# Bereken de 'Gem Fossil' ratio en voeg deze toe als nieuwe kolom aan merged_df
# Let op de correcte kolomnamen uit de merged_df voor de berekening.
# Zorg ervoor dat 'Fossil fuels per capita' en 'Population' kolommen bestaan in merged_df.
merged_df['Gem Fossil'] = merged_df['Fossil fuels per capita'] / merged_df['Population']

print("\nSamengevoegde dataframe met 'Gem Fossil' kolom (eerste 5 rijen):")
display(merged_df.head())
```

```python
merged_df.drop(['Code_x', 'Code_y'], axis=1, inplace=True)
print("Kolommen 'Code_x' en 'Code_y' verwijderd.")
```

```python
import matplotlib.pyplot as plt

# Filter out the rows that aren't countries and reset the index
df_filtered = fossil_fuels_df[fossil_fuels_df['Entity'].str.contains("World|European Union") == False].reset_index(drop=True)

# Select only the relevant years for the calculation
years_for_pivot = [1990, 2005, 2017, 2021] # Years as integers
df_calculation = df_filtered[df_filtered['Year'].isin(years_for_pivot)]

# Pivot the table so that years become columns and 'Fossil fuels per capita' are the values
merged_df = df_calculation.pivot_table(index='Entity', columns='Year', values='Fossil fuels per capita').reset_index()

# Convert year column names from integers to strings to match 'years' list
merged_df.columns = [str(col) if isinstance(col, int) else col for col in merged_df.columns]
merged_df.columns.name = None # Remove the 'Year' name from columns index

# Calculate the relative change
years = ['1990', '2005', '2017', '2021']
for year in years:
    rel_column = f"rel_{year}"
    # Handle potential division by zero by replacing 0 with NaN or a small number
    # For simplicity, we assume '1990' values are non-zero or Inf/NaN will be handled by pandas.
    merged_df[rel_column] = merged_df[year] / merged_df['1990'] * 100

# Since we are dividing by '1990', make sure it's not zero to avoid division by zero errors
merged_df['rel_1990'] = 100

# Copy dataframe for the next exercise to prevent having to do the same calculation twice
merged_df_5mt = merged_df.copy()

# Find the smallest and largest changers
df_largest_changers = merged_df.nlargest(3, 'rel_2021')
df_smallest_changers = merged_df.nsmallest(3, 'rel_2021')

# Set the country column as index, to make plotting more easy
df_largest_changers.set_index('Entity', inplace=True)
df_smallest_changers.set_index('Entity', inplace=True)

# Rename column names to beautify the figure
new_column_names = {'rel_1990': '1990', 'rel_2005': '2005', 'rel_2017': '2017', 'rel_2021': '2021'}
df_smallest_changers.rename(columns=new_column_names, inplace=True)
df_largest_changers.rename(columns=new_column_names, inplace=True)


# Plot
plt.figure(figsize=[10,8])

for index, row in df_largest_changers.iterrows():
    plt.plot(row.index, row.values, marker='o', label=index)

for index, row in df_smallest_changers.iterrows():
    plt.plot(row.index, row.values, marker='o', label=index)


plt.title('The Relative % Carbon Dioxide Emissions Over Time')
plt.xlabel('Year')
plt.ylabel('Relative CO2 Emissions %')
plt.legend()
# plt.yscale('log')
plt.grid()
plt.show()
```
