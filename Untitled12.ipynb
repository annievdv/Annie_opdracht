```python
import numpy as np
import pandas as pd
from google.colab import files

# Lees de fossiele brandstoffen data
fossil_fuels_df = pd.read_csv('https://raw.githubusercontent.com/annievdv/Annie_opdracht/refs/heads/main/fossil-fuels-per-capita.csv')
fossil_fuels_df.rename(columns={'Entity': 'Land'}, inplace=True)
print("Fossiele brandstoffen data:")
display(fossil_fuels_df)

# Filter out 'World' and 'European Union' from df_filtered initially
# Assuming df_filtered is defined from previous cells and contains 'Entity', 'Year', 'Fossil fuels per capita'
df_filtered_countries = fossil_fuels_df[fossil_fuels_df['Land'].str.contains("World|European Union") == False].reset_index(drop=True)

# Lees de bevolkingsdata
population_df = pd.read_csv('https://raw.githubusercontent.com/annievdv/Annie_opdracht/refs/heads/main/population-with-un-projections.csv')
population_df.rename(columns={'Entity': 'Land'}, inplace=True)

print("\nBevolkingsdata ")
display(population_df)

# Filter out 'World' and 'European Union' from df_filtered initially
# Assuming df_filtered is defined from previous cells and contains 'Entity', 'Year', 'population-with-un-projections'
df_filtered_countries2 = population_df[population_df['Land'].str.contains("World|European Union") == False].reset_index(drop=True)

# Aangenomen dat fossil_fuels_df en population_df nu correct zijn gedefinieerd in de vorige cel.
# Samenvoegen op gemeenschappelijke kolommen 'Land' en 'Year'.
merged_df = ""
merged_df = pd.merge(fossil_fuels_df, population_df, on=['Land', 'Year'], how='outer')
print("Samengevoegde dataframe")
display(merged_df)
```

```python
# Bereken de 'Gem Fossil' ratio en voeg deze toe als nieuwe kolom aan merged_df
# Let op de correcte kolomnamen uit de merged_df voor de berekening.
# Zorg ervoor dat 'Fossil fuels per capita' en 'Population' kolommen bestaan in merged_df.
merged_df['Gem Fossil'] = merged_df['Fossil fuels per capita'] / merged_df['Population']

print("\nSamengevoegde dataframe met 'Gem Fossil' kolom ")
display(merged_df)
```

```python
import matplotlib.pyplot as plt
import pandas as pd

# Make a copy of merged_df to preserve the original, if needed later.
# This assumes 'merged_df' is already defined in a previous cell.
df_changers = merged_df.copy()

# The original 'years' list contains string representations of years.
# We need to use integer years for filtering the 'Year' column.
years = ['1990', '2005', '2017', '2021']
years_int = [int(y) for y in years] # Convert to int for filtering

# Filter merged_df for the relevant years and columns needed for pivoting.
# This creates a temporary DataFrame in 'long' format.
df_temp_pivot = df_changers[df_changers['Year'].isin(years_int)][['Land', 'Year', 'Fossil fuels per capita']].copy()

# Pivot the filtered DataFrame.
# 'Land' becomes the index, 'Year' becomes columns, and 'Fossil fuels per capita' are the values.
# If a country has multiple entries for a year, pivot_table will average them by default.
df_changers = df_temp_pivot.pivot_table(index='Land', columns='Year', values='Fossil fuels per capita')

# After pivoting, the 'Year' column name might still be present as columns.name. Remove it.
df_changers.columns.name = None
# Make 'Land' a regular column again from the index.
df_changers.reset_index(inplace=True)

# Rename the year columns (which are currently integers from pivot_table) to strings,
# to align with the 'years' list and original code's intent of using string years for column access.
df_changers.rename(columns={y_int: str(y_int) for y_int in years_int}, inplace=True)

# Calculate the relative change for the years
# Initialize 'rel_' columns to NaN.
for year_str in years:
    df_changers[f"rel_{year_str}"] = float('nan')

# Identify rows where '1990' data is valid for baseline calculation (not NaN and not zero).
valid_1990_baseline_mask = df_changers['1990'].notna() & (df_changers['1990'] != 0)

# Apply calculations only to rows with a valid 1990 baseline.
if valid_1990_baseline_mask.any():
    for year_str in years:
        rel_column = f"rel_{year_str}"
        # Calculate relative change. loc ensures we only modify rows with valid 1990 baseline.
        df_changers.loc[valid_1990_baseline_mask, rel_column] = \
            (df_changers.loc[valid_1990_baseline_mask, year_str] / \
             df_changers.loc[valid_1990_baseline_mask, '1990']) * 100

    # As per original code's intent, explicitly set 'rel_1990' to 100 for valid baselines.
    df_changers.loc[valid_1990_baseline_mask, 'rel_1990'] = 100

# Copy dataframe for the next exercise to prevent having to do the same calculation twice
df_changers_5mt = df_changers.copy()

# Drop the original year columns ('1990', '2005', '2017', '2021') that were created during pivoting.
# This aligns with the original code's `df_changers.drop(columns=years, inplace=True)`.
df_changers.drop(columns=years, inplace=True)

# Find the smallest and largest changers
# 'Land' is now a column after reset_index, so it can be used directly for nlargest.
df_smallest_changers = df_changers.nsmallest(10, 'rel_2021')

# Set the 'Land' column as index for plotting, consistent with the original code's `set_index('Country', inplace=True)`.
# Using 'Land' instead of 'Country' as per the DataFrame's actual column name.
df_smallest_changers.set_index('Land', inplace=True)

# Get the list of top 10 country names
top_10_countries = df_smallest_changers.index.tolist() # Get index which is now 'Land'

# Plot
plt.figure(figsize=[12,8]) # Increased figure size for better readability

for country in top_10_countries:
    country_data = merged_df[merged_df['Land'] == country] # Use merged_df for plotting
    plt.plot(country_data['Year'], country_data['Fossil fuels per capita'], label=country)


plt.title('Top 10 Countries by Fossil Fuel Emissions Per Capita')
plt.xlabel('Year')
plt.ylabel('Fossil fuels per capita')
plt.xlim(1990,2024)
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left') # Place legend outside the plot
plt.grid(True) # Changed from plt.grid() to plt.grid(True) for clarity
plt.tight_layout() # Adjust layout to prevent labels from overlapping
plt.show()

df_smallest_changers
```
