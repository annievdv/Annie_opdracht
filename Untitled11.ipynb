```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from google.colab import files

# Lees de data
df=pd.read_csv('https://raw.githubusercontent.com/annievdv/Annie_opdracht/refs/heads/main/levelized-cost-of-energy.csv')

print("cost data ")
display(df.head())

```

```python
df_filtered = df[df['Entity'].str.contains("World|European Union") == False].reset_index(drop=True)

xOnshoreWind = 'Onshore wind'
xBioenergy = 'Bioenergy'
xOffshoreWind = 'Offshore wind'
xGeothermal = 'Geothermal'
xSolar = 'Solar'
xConcent = 'Concentrated solar power'
xHydro = 'Hydropower'

# Filter df_filtered for the year 2024 to find the largest emitters in that year
df_for_ranking = df_filtered[df_filtered['Year'] == 2024]
df_for_ranking = df_for_ranking.dropna(subset=['Entity', xOnshoreWind]) # Using xOnshoreWind for ranking

largest_countries = df_for_ranking.nlargest(10, xOnshoreWind)

# Get the list of top 10 country names
top_10_countries = largest_countries['Entity'].tolist()

# Filter the original df_filtered (containing all years, but no World/EU) for these top 10 countries
df_plot = df_filtered[df_filtered['Entity'].isin(top_10_countries)]

# definieer de x en y-assen voor de lineaire regressie voor Onshore Wind
x_onshore = df_plot['Year']
y_onshore = df_plot[xOnshoreWind]

yearsToPredict = np.arange(1990,2051)

# --- Regression for Onshore Wind ---
# Before fitting, drop rows where x_onshore or y_onshore contain NaN values
df_temp_onshore = pd.DataFrame({'Year': x_onshore, 'Onshore wind': y_onshore}).dropna()
x_cleaned_onshore = df_temp_onshore['Year']
y_cleaned_onshore = df_temp_onshore['Onshore wind']

# Fit the model for Onshore Wind using the cleaned data
model_onshore = np.polynomial.Polynomial.fit(x_cleaned_onshore, y_cleaned_onshore, deg=1)
pricePredictions_onshore = model_onshore(yearsToPredict)

# --- Regression for Geothermal ---
# definieer de x en y-assen voor de lineaire regressie voor Geothermal
x_geothermal = df_plot['Year']
y_geothermal = df_plot[xGeothermal] # Use geothermal here

# Before fitting, drop rows where x_geothermal or y_geothermal contain NaN values
df_temp_geothermal = pd.DataFrame({'Year': x_geothermal, 'Geothermal': y_geothermal}).dropna()
x_cleaned_geothermal = df_temp_geothermal['Year']
y_cleaned_geothermal = df_temp_geothermal['Geothermal']

# Initialize pricePredictions_geothermal to None
pricePredictions_geothermal = None

# Fit the model for Geothermal using the cleaned data, only if there's actual data
if not df_temp_geothermal.empty:
    model_geothermal = np.polynomial.Polynomial.fit(x_cleaned_geothermal, y_cleaned_geothermal, deg=1)
    pricePredictions_geothermal = model_geothermal(yearsToPredict)
else:
    print(f"Warning: No valid data found for '{xGeothermal}' to perform regression for the selected countries.")

# --- Regression for Offshore wind ---
# definieer de x en y-assen voor de lineaire regressie voor Offshore wind
x_offshorewind = df_plot['Year']
y_offshorewind = df_plot[xOffshoreWind] # Use xOffshoreWind here

# Before fitting, drop rows where x_offshorewind or y_offshorewind contain NaN values
df_temp_offshorewind = pd.DataFrame({'Year': x_offshorewind, 'Offshore wind': y_offshorewind}).dropna()
x_cleaned_offshorewind = df_temp_offshorewind['Year']
y_cleaned_offshorewind = df_temp_offshorewind['Offshore wind']

# Initialize pricePredictions_offshorewind to None
pricePredictions_offshorewind = None

# Fit the model for offshorewind using the cleaned data, only if there's actual data
if not df_temp_offshorewind.empty:
    model_offshorewind = np.polynomial.Polynomial.fit(x_cleaned_offshorewind, y_cleaned_offshorewind, deg=1)
    pricePredictions_offshorewind = model_offshorewind(yearsToPredict)
else:
    print(f"Warning: No valid data found for '{xOffshoreWind}' to perform regression for the selected countries.")


# --- Regression for Hydropower ---
pricePredictions_hydro = None # Initialize outside conditional block

if xHydro in df_plot.columns: # Check if 'Hydropower' column exists
    # definieer de x en y-assen voor de lineaire regressie voor Hydropower
    x_hydro = df_plot['Year']
    y_hydro = df_plot[xHydro]

    # Before fitting, drop rows where x_hydro or y_hydro contain NaN values
    df_temp_hydro = pd.DataFrame({'Year': x_hydro, xHydro: y_hydro}).dropna() # Corrected column name to xHydro
    x_cleaned_hydro = df_temp_hydro['Year']
    y_cleaned_hydro = df_temp_hydro[xHydro] # Corrected column name to xHydro

    # Fit the model for Hydropower using the cleaned data, only if there's actual data
    if not df_temp_hydro.empty:
        model_hydro = np.polynomial.Polynomial.fit(x_cleaned_hydro, y_cleaned_hydro, deg=1)
        pricePredictions_hydro = model_hydro(yearsToPredict)
    else:
        print(f"Warning: No valid data found for '{xHydro}' to perform regression for the selected countries.")
else:
    print(f"Warning: Column '{xHydro}' not found in the data. Skipping regression for {xHydro}.")


# --- Regression for Solar ---
pricePredictions_solar = None # Initialize outside conditional block

if xSolar in df_plot.columns: # Check if 'Solar' column exists
    # definieer de x en y-assen voor de lineaire regressie voor Solar
    x_solar = df_plot['Year']
    y_solar = df_plot[xSolar]

    # Before fitting, drop rows where x_solar or y_solar contain NaN values
    df_temp_solar = pd.DataFrame({'Year': x_solar, 'Solar': y_solar}).dropna()
    x_cleaned_solar = df_temp_solar['Year']
    y_cleaned_solar = df_temp_solar['Solar']

    # Fit the model for Solar using the cleaned data, only if there's actual data
    if not df_temp_solar.empty:
        model_solar = np.polynomial.Polynomial.fit(x_cleaned_solar, y_cleaned_solar, deg=1)
        pricePredictions_solar = model_solar(yearsToPredict)
    else:
        print(f"Warning: No valid data found for '{xSolar}' to perform regression for the selected countries.")
else:
    print(f"Warning: Column '{xSolar}' not found in the data. Skipping regression for {xSolar}.")


# --- Regression for Concentrated Solar power ---
pricePredictions_concent = None # Initialize outside conditional block

if xConcent in df_plot.columns: # Check if 'Concentrated solar power' column exists
    # definieer de x en y-assen voor de lineaire regressie voor Concentrated solar power
    x_concent = df_plot['Year']
    y_concent = df_plot[xConcent]

    # Before fitting, drop rows where x_concent or y_concent contain NaN values
    df_temp_concent = pd.DataFrame({'Year': x_concent, xConcent: y_concent}).dropna()
    x_cleaned_concent = df_temp_concent['Year']
    y_cleaned_concent = df_temp_concent[xConcent]

    # Fit the model for Concentrated solar power using the cleaned data, only if there's actual data
    if not df_temp_concent.empty:
        model_concent = np.polynomial.Polynomial.fit(x_cleaned_concent, y_cleaned_concent, deg=1)
        pricePredictions_concent = model_concent(yearsToPredict)
    else:
        print(f"Warning: No valid data found for '{xConcent}' to perform regression for the selected countries.")
else:
    print(f"Warning: Column '{xConcent}' not found in the data. Skipping regression for {xConcent}.")


# --- Regression for Bioenergy ---
# definieer de x en y-assen voor de lineaire regressie voor Bioenergy
x_bio = df_plot['Year']
y_bio = df_plot[xBioenergy] # Use xBioenergy here

# Before fitting, drop rows where x_bio or y_bio contain NaN values
df_temp_bio = pd.DataFrame({'Year': x_bio, 'Bioenergy': y_bio}).dropna()
x_cleaned_bio = df_temp_bio['Year']
y_cleaned_bio = df_temp_bio['Bioenergy']

# Initialize pricePredictions_bio to None
pricePredictions_bio = None

# Fit the model for Bioenergy using the cleaned data, only if there's actual data
if not df_temp_bio.empty:
    model_bio = np.polynomial.Polynomial.fit(x_cleaned_bio, y_cleaned_bio, deg=1)
    pricePredictions_bio = model_bio(yearsToPredict)
else:
    print(f"Warning: No valid data found for '{xBioenergy}' to perform regression for the selected countries.")


# Plotting the data over time
plt.figure(figsize=[12, 8])
for country in top_10_countries:
   country_data = df_plot[df_plot['Entity'] == country]
   # Plot actual data for Onshore Wind
   plt.plot(country_data['Year'], country_data[xOnshoreWind], label=f'{country} Onshore Wind', marker='o', markersize=4)
   # Plot actual data for Bioenergy (if not NaN)
   if xBioenergy in country_data.columns and not country_data[xBioenergy].isnull().all(): # Added column check
       plt.plot(country_data['Year'], country_data[xBioenergy], label=f'{country} Bioenergy', marker='x', markersize=4, linestyle=':')
   # Plot actual data for Geothermal (if not NaN)
   if xGeothermal in country_data.columns and not country_data[xGeothermal].isnull().all(): # Added column check
       plt.plot(country_data['Year'], country_data[xGeothermal], label=f'{country} Geothermal', marker='D', markersize=4, linestyle='--') # Changed marker and linestyle for Geothermal
   # Plot actual data for Offshore Wind (if not NaN)
   if xOffshoreWind in country_data.columns and not country_data[xOffshoreWind].isnull().all(): # Added column check
       plt.plot(country_data['Year'], country_data[xOffshoreWind], label=f'{country} Offshore Wind', marker='s', markersize=4, linestyle='-.') # Changed marker and linestyle for Offshore Wind
   # Plot actual data for Solar (if not NaN)
   if xSolar in country_data.columns and not country_data[xSolar].isnull().all(): # Added column check
       plt.plot(country_data['Year'], country_data[xSolar], label=f'{country} Solar', marker='*', markersize=4, linestyle=':') # Changed marker and linestyle for Solar
   # Plot actual data for Concentrated Solar power (if not NaN)
   if xConcent in country_data.columns and not country_data[xConcent].isnull().all(): # Added column check
       plt.plot(country_data['Year'], country_data[xConcent], label=f'{country} Concentrated Solar', marker='P', markersize=4, linestyle=':') # Added plot for Concentrated Solar
   # Plot actual data for Hydropower (if not NaN)
   if xHydro in country_data.columns and not country_data[xHydro].isnull().all(): # Added column check
       plt.plot(country_data['Year'], country_data[xHydro], label=f'{country} Hydropower', marker='^', markersize=4, linestyle='-') # Added plot for Hydropower


# Plot the linear regression predictions for Onshore Wind
plt.plot(yearsToPredict, pricePredictions_onshore, color='red', linestyle='--', label='Onshore Wind Regressie')

# Plot the linear regression predictions for Bioenergy, only if calculated
if pricePredictions_bio is not None:
    plt.plot(yearsToPredict, pricePredictions_bio, color='green', linestyle='--', label='Bioenergy Regressie')

# Plot the linear regression predictions for Geothermal, only if calculated
if pricePredictions_geothermal is not None:
    plt.plot(yearsToPredict, pricePredictions_geothermal, color='blue', linestyle='--', label='Geothermal Regressie') # Updated label

# Plot the linear regression predictions for Offshore Wind, only if calculated
if pricePredictions_offshorewind is not None:
    plt.plot(yearsToPredict, pricePredictions_offshorewind, color='purple', linestyle='--', label='Offshore Wind Regressie')

# Plot the linear regression predictions for Solar, only if calculated
if pricePredictions_solar is not None:
    plt.plot(yearsToPredict, pricePredictions_solar, color='orange', linestyle='--', label='Solar Regressie')

# Plot the linear regression predictions for Concentrated Solar, only if calculated
if pricePredictions_concent is not None:
    plt.plot(yearsToPredict, pricePredictions_concent, color='brown', linestyle='--', label='Concentrated Solar Regressie') # Added plot for Concentrated Solar Regressie

# Plot the linear regression predictions for Hydropower, only if calculated
if pricePredictions_hydro is not None:
    plt.plot(yearsToPredict, pricePredictions_hydro, color='cyan', linestyle='--', label='Hydropower Regressie') # Added plot for Hydropower Regressie

# Add labels and a title
plt.xlabel('Year')
plt.ylabel('Energy Production (e.g., in TWh or similar units)') # More general label
plt.title('Niet-fossiele brandstofproductie en prognoses Top 10 Landen') # Updated title for clarity

plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.grid(True)
plt.show()
```
